1. Read Raw Data .xlsx, then take some predetermined columns

```python
import pandas as pd 

path = "D:/Personal/Project/World Energy/Dataset/Raw Data/owid-energy-data.xlsx"
df = pd.read_excel(path, usecols=["country", "iso_code", "year", "population", "gdp", "wind_electricity", "wind_elec_per_capita", "wind_consumption"])

df.head()
```

2. Set year 1974 - 2024

```python
df = df[(df['year'] >= 1974) & (df['year'] <= 2024)]

df.head()
```

3. Handle missing value and for feature "population" & "gdp", interpolate and fill the data.

```python
def clean_data(df):
    # Drop rows with missing data in column: 'iso_code'
    df = df.dropna(subset=['iso_code'])
    df[['population', 'gdp']] = (
                    df.groupby('country')[['population', 'gdp']]
                    .transform(lambda x: x.interpolate())
                    .bfill()
                    .ffill()
                )
    # Replace missing values with 0 in column: 'wind_consumption'
    df = df.fillna({'wind_consumption': 0})
    # Replace missing values with 0 in column: 'wind_elec_per_capita'
    df = df.fillna({'wind_elec_per_capita': 0})
    # Replace missing values with 0 in column: 'wind_electricity'
    df = df.fillna({'wind_electricity': 0})
    return df

df_clean = clean_data(df.copy())
df_clean.head()
```

4. Export to excel file

```python
df_clean.to_excel("clean_data_wind.xlsx", index=False)

print("File completed saved!")
```
