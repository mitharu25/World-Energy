1. Read Raw Data .xlsx, then take column country, iso_code, year, and population

```python
import pandas as pd 

path = "D:/Personal/Project/World Energy/Dataset/Raw Data/owid-energy-data.xlsx"
df = pd.read_excel(path, usecols=["country", "iso_code", "year", "population"])

df.head()
```

2. Set year 1974 - 2024

```python
df = df[(df['year'] >= 1974) & (df['year'] <= 2024)]

df.head()
```

3. Handle missing value

```python
def clean_data(df):
    # Drop rows with missing data in column: 'iso_code'
    df = df.dropna(subset=['iso_code'])
    # Replace gaps forward from the previous valid value in: 'population'
    df = df.fillna({'population': df['population'].ffill()})
    return df

df_clean = clean_data(df.copy())
df_clean.head()
```

4. Export to excel file

```python
df_clean.to_excel("clean_population.xlsx", index=False)

print("File completed saved!")
```
